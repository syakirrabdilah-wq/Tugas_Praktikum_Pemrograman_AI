```python
print("Praktikum Pemrograman Artificial Inteligence")

# Boolean
status = False
print(status)
print(type(status))

# String
teks = "Ini adalah tulisan berupa String"
print(teks)

# Integer
bilangan_bulat = 100
print(bilangan_bulat)

# Float
bilangan_desimal = 0.001
print(bilangan_desimal)

# Bilangan heksadesimal dan biner
print("bilangan desimal dari 0x01 adalah", 0x01)
print(0b1010)

# Complex
kompleks_1 = 1j
print(type(kompleks_1))
kompleks_2 = 2 + 6j
print(kompleks_2)
print(type(kompleks_2))

# List
angka = [96, 97, 98, 99, 100]
print(angka)
teks_list = ["seratus", "dua ratus", "tiga ratus"]
print(teks_list)

# Dictionary
biodata = {"nama": "Syakir", "umur": 19}
print(biodata)

# String dengan tanda kutip tunggal
print('Teknik Informatika')
```
