Pengulangan While Loop di dalam bahasa pemrograman Python dieksesusi statement berkali-kali selama kondisi bernilai benar atau True.

Dibawah ini adalah contoh penggunaan pengulangan While Loop

```python
#Contoh penggunaan While Loop

count = 0
while (count < 9):
    print ("The count is: ", count)
    count = count + 1

print ("Good bye!")
```

```python
x = "FASILKOM"
while x:
    print(x, " ")
    x = x[1:]
```

Pengulangan for pada Python memiliki kemampuan untuk mengulangi item dari urutan apapun, seperti list atau string.

Dibawah ini adalah contoh penggunaan pengulangan For Loop.

```python
#Contoh pengulangan for sederhana
angka = [1,2,3,4,5]
for x in angka:
    print(x)

#Contoh pengulangan for
buah = ["nanas", "apel", "jeruk"]
for makanan in buah:
    print ("Saya suka makan", makanan)
```

```python
nama = ['budi', 'andi', 'rudi', 'sandi'] 
usia = [20, 18, 22, 19] 
for i in range(len(nama)) : 
    print(nama[i], ' berusia ', usia[i], ' tahun')
```

Bahasa pemrograman Python memungkinkan penggunaan satu lingkaran di dalam loop lain. Bagian berikut menunjukkan beberapa contoh untuk menggambarkan konsep tersebut.

Dibawah ini adalah contoh penggunaan Nested Loop.

```python
#Contoh penggunaan Nested Loop

i = 2
while(i < 100):
    j = 2
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if (j > i/j) : print(i, " is prime")
    i = i + 1

print("Good bye!")
```

```python
# Program 1: Memfilter bilangan genap dan ganjil

angka = [3, 8, 12, 7, 20, 15, 4, 9, 10, 1]

genap = []
ganjil = []

# Memisahkan bilangan genap dan ganjil
for x in angka:
    if x % 2 == 0:
        genap.append(x)
    else:
        ganjil.append(x)

# Menghitung jumlah (total) masing-masing
total_genap = 0
for x in genap:
    total_genap = total_genap + x

total_ganjil = 0
for x in ganjil:
    total_ganjil = total_ganjil + x

print("List awal       :", angka)
print("Bilangan genap  :", genap)
print("Bilangan ganjil :", ganjil)
print("Banyak genap    :", len(genap))
print("Banyak ganjil   :", len(ganjil))
print("Jumlah genap    :", total_genap)
print("Jumlah ganjil   :", total_ganjil)
```

```python
# Program 2: Menampilkan perkalian dari angka yang dimasukkan

angka = int(input("Masukkan angka: "))
batas = int(input("Sampai perkalian ke berapa? "))

i = 1
while i <= batas:
    hasil = angka * i
    print(angka, "x", i, "=", hasil)
    i = i + 1
```

```python
# Program 3: Menampilkan deret Fibonacci

n = int(input("Masukkan banyak suku Fibonacci: "))

a = 0
b = 1
hitung = 0

print("Deret Fibonacci:")
while hitung < n:
    print(a, end=" ")
    c = a + b   # suku berikutnya = jumlah dua suku sebelumnya
    a = b
    b = c
    hitung = hitung + 1
print()
```

```python
# Program 4: Tabel perkalian 1-10 (hasil sampai 100) dengan nested loop

for baris in range(1, 11):
    for kolom in range(1, 11):
        hasil = baris * kolom
        print(hasil, end="\t")   # \t = jarak tab agar rapi
    print()   # pindah ke baris baru
```
